```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/processed/feature_engineered_data.csv"
)

df.head()
```

```python
df.groupby("country").size()
```

```python
df.groupby("country").size().sort_values(
    ascending=False
).head(10)
```

```python
country_quantity = (
    df.groupby("country")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

country_quantity.head(10)
```

```python
country_analysis = (
    df.groupby("country")
    .agg(
        total_quantity=("quantity", "sum"),
        average_quantity=("quantity", "mean"),
        average_price=("unitprice", "mean"),
        transactions=("invoiceno", "nunique")
    )
    .sort_values(
        "total_quantity",
        ascending=False
    )
)

country_analysis.head(10)
```

```python
country_analysis.sort_values(
    "transactions",
    ascending=False
).head(10)
```

```python
country_analysis.sort_values(
    "transactions",
    ascending=True
).head(10)
```

```python
pivot = pd.pivot_table(
    df,
    values="quantity",
    index="country",
    aggfunc="sum"
)

pivot.sort_values(
    "quantity",
    ascending=False
).head(10)
```

```python
df.columns.tolist()
```

```python
monthly_quantity = (
    df.groupby(
        ["InvoiceDate_year", "InvoiceDate_month"]
    )["quantity"]
    .sum()
)

monthly_quantity
```

```python
country_analysis["rank"] = (
    country_analysis["total_quantity"]
    .rank(
        ascending=False,
        method="dense"
    )
)

country_analysis.head(10)
```

```python
total_quantity = df["quantity"].sum()
```

```python
country_analysis["contribution_percent"] = (
    country_analysis["total_quantity"]
    / total_quantity
    * 100
)
```
